```python
# Catagory the people by their age like children, adult, citizen, senior citizen..
```

```python
i = int(input("age:"))
if i < 18:
    print ("Children")
elif 18 <= i <= 35:
    print ("Adult")
elif 36 <= i <= 59:
    print ("Citizen")  
else:
    print ("Senior Citizen")
```
