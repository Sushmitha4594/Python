```python
bmi = int(input("Enter the BMI Index:"))
if bmi <18:
    print ("Underweight")
elif 18 < bmi < 24:
    print ("Normal weight")
elif 24 < bmi < 28:
    print ("Overweight")
elif 28 < bmi < 35:
    print ("Very Overweight")
else:
    print ("Extermely Overweight (Obese)")
```
